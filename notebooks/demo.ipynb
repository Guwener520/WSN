# WSN Coverage Optimization - Demo
Demonstrates the core workflow: environment setup, algorithm optimization, and visualization.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
from wsn.environment import WSNEnvironment
from wsn.fitness import FitnessEvaluator
from wsn.algorithms import PSO, ImprovedPSO, GWO, ImprovedGWO, WOA, ImprovedWOA
from wsn.experiments import RandomBaseline, GridBaseline, ExperimentRunner

print('All imports successful')
```

```python
# Create environment: 100x100 area, 50 nodes, sensing radius 12
env = WSNEnvironment(
    width=100, height=100,
    grid_resolution=2.0,
    n_nodes=50,
    sensing_radius=12.0,
    communication_radius=24.0,
    seed=42,
)

evaluator = FitnessEvaluator(w_coverage=0.5, w_redundancy=0.3, w_connectivity=0.2)
print(f'Grid points: {env.n_grid_points}, Area: {env.width}x{env.height}')
```

```python
# Run PSO optimization
env_pso = WSNEnvironment(width=100, height=100, grid_resolution=2.0, n_nodes=50,
                          sensing_radius=12.0, communication_radius=24.0, seed=42)
pso = PSO(env_pso, evaluator, n_particles=30, max_iter=100, seed=42)
best_pos, best_fit = pso.optimize(verbose=True)
print(f'\nPSO best fitness: {best_fit:.4f}')
```

```python
# Visualize result
from wsn.visualization import plot_deployment, plot_coverage_heatmap
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
plot_deployment(env_pso, best_pos, ax=ax1, title='PSO Optimized Deployment')
plot_coverage_heatmap(env_pso, best_pos, ax=ax2, title='Coverage Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Convergence curve
from wsn.visualization import plot_convergence
plot_convergence([pso.history], ['PSO'], title='PSO Convergence')
plt.show()
```
